```python
import numpy as np
import pyNN.spiNNaker as sim
import random
import matplotlib.pyplot as plt
```

```python
class Tictactoe_v0:
    def __init__(self):
        self.board = [0] * 9
        self.wining_position = [[0, 1, 2], [3, 4, 5], [6, 7, 8],
                                [0, 3, 6], [1, 4, 7], [2, 5, 8],
                                [0, 4, 8], [6, 4, 2]]
        self.current_turn = 1
        self.player_mark = 1

    def reset(self, is_human_first=True):
        self.board = [0] * 9
        self.current_turn = 1
        self.player_mark = 1 if is_human_first else -1
#         # uncomment the below
#         print("\n--- New Episode ---")
#         self.render() 
        if not is_human_first:
            self.env_act()
        return self.board.copy()

    def check_win(self):
        for pst in self.wining_position:
            if str(self.board[pst[0]]) + str(self.board[pst[1]]) + str(self.board[pst[2]]) in ['111', '-1-1-1']:
                if self.current_turn == self.player_mark:
                    return 1, True
                return -1, True
        if 0 not in self.board:
            return 0, True
        return 0, False

    def env_act(self):
        action = random.choice([i for i in range(len(self.board)) if self.board[i] == 0])
        for pst in self.wining_position:
            com = str(self.board[pst[0]]) + str(self.board[pst[1]]) + str(self.board[pst[2]])
            if com.replace('0', '') == str(self.current_turn) * 2:
                if self.board[pst[0]] == 0:
                    action = pst[0]
                elif self.board[pst[1]] == 0:
                    action = pst[1]
                else:
                    action = pst[2]
        if self.board[action] != 0:
            raise Exception('Invalid action')
        self.board[action] = self.current_turn
        reward, done = self.check_win()
        self.current_turn = self.current_turn * -1
#         # uncomment the below
#         self.render()
        return reward, done
    
    def step(self, action):
        """Handle invalid moves gracefully"""
        valid_actions = self.get_valid_actions()
        
        # Auto-correct invalid actions
        if action not in valid_actions:
            action = random.choice(valid_actions) if valid_actions else None
            
        if action is not None:    
            self.board[action] = self.current_turn
#             # uncomment the below
#             self.render()
            reward, done = self.check_win()
            self.current_turn *= -1
            
            if not done:
                reward, done = self.env_act()
#                 # uncomment the below
#                 self.render()
        else:
            # No valid moves left (should never happen with proper checks)
            reward, done = 0, True
            
        return self.board.copy(), reward, done, None

#     def step(self, action):
#         if self.board[action] != 0:
#             raise Exception('Invalid action')
#         self.board[action] = self.current_turn
#         # uncomment the below
#         self.render()
#         reward, done = self.check_win()
#         self.current_turn = self.current_turn * -1
#         # if done:
#         #     return self.board.copy(), reward, done, None
#         # reward, done = self.env_act()
#         # return self.board.copy(), reward, done, None
#         if not done:
#             reward, done = self.env_act()
#             # uncomment the below
# #             self.render()  # Render after O move
#         return self.board.copy(), reward, done, None
    
    
    def get_valid_actions(self):
        return [i for i, val in enumerate(self.board) if val == 0]
    
    
#     #uncomment the below
#     def render(self):
#         """Visualize the board with X/O positions"""
#         symbols = {1: 'X', -1: 'O', 0: ' '}
#         print("\nCurrent Board:")
#         for i in range(3):
#             print(f" {symbols[self.board[i*3]]} | {symbols[self.board[i*3+1]]} | {symbols[self.board[i*3+2]]} ")
#             if i < 2: print("-----------")
#         print()
```

```python
def population_encode(state):
    """Encodes Tic-Tac-Toe board state using population encoding."""
    encoded = np.zeros((9, 3))
    for i, val in enumerate(state):
        if val == 1:
            encoded[i] = [0.1, 0.2, 0.9]  # X (Fire third neuron)
        elif val == -1:
            encoded[i] = [0.9, 0.2, 0.1]  # O (Fire first neuron)
        else:
            encoded[i] = [0.2, 0.8, 0.2]  # Empty (Fire middle neuron)
    #uncomment the below
#     print("Flattened encoded vector:", encoded.flatten())
    return encoded.flatten()

# def decode_action(spike_data):
#     """Decode action based on the neuron with the highest spike count."""
#     spike_counts = [len(spikes) for spikes in spike_data]
#     return np.argmax(spike_counts)

def decode_action_by_potential(output_pop, valid_actions):
    """
    Decode action based on membrane potential of output neurons.
    Only considers valid actions.
    """
    potentials = output_pop.get_data("v").segments[-1].filter(name='v')[0][-1].magnitude
    
#     print("\n=== Action Decoding (by potential) ===")
#     for i, pot in enumerate(potentials):
#         print(f"Action {i}: {pot:.2f} mV")

    valid_potentials = [(i, pot) for i, pot in enumerate(potentials) if i in valid_actions]
    best_action = max(valid_potentials, key=lambda x: x[1])[0]
    
#     print(f"Selected action: {best_action}")
    return best_action

# def decode_action(spike_data):
#     """Decode action based on the neuron with the highest spike count."""
#     spike_counts = [len(spikes) for spikes in spike_data]
#     print("\n=== Action Decoding ===")
#     print("Output neuron spike counts:", spike_counts)
    
#     max_count = max(spike_counts)
#     candidates = [i for i, count in enumerate(spike_counts) if count == max_count]
#     action = random.choice(candidates) if len(candidates) > 1 else candidates[0]
    
#     print(f"Selected action: {action} ({'tie break' if len(candidates)>1 else 'clear winner'})")
#     return action
```

```python
# -------------------------------
# Reward-Modulated Per-Synapse STDP
# -------------------------------
def stdp_per_synapse_update(
    current_weights,
    reward,
    input_spike_trains,
    hidden_spike_trains,
    run_offset,
    step_size=10.0,  # The ms we run for each move
    tau_plus=20.0,
    tau_minus=20.0,
    A_plus=0.05,
    A_minus=0.05,
):
    """
    For each pair (pre, post), compute weight update using spike times in the
    most recent time window [run_offset, run_offset+step_size].
    Reward scales the STDP magnitude.
    """
    if reward == 0.0:
        # If you only want to update on non-zero rewards, skip here
        return current_weights

    # For each synapse (pre_idx -> post_idx)
    #loop through each neuron and record spike times for that input neuron
    for pre_idx in range(current_weights.shape[0]):
        pre_times_all = input_spike_trains[pre_idx]
        # Filter pre times to only the last step window
        pre_times_step = pre_times_all[
            (pre_times_all >= run_offset) & (pre_times_all < run_offset + step_size)
        ]
        #loop through each postsynaptic neuron and record spikes for the output neuron
        for post_idx in range(current_weights.shape[1]):
            post_times_all = hidden_spike_trains[post_idx]
            # Filter post times to last step
            post_times_step = post_times_all[
                (post_times_all >= run_offset) & (post_times_all < run_offset + step_size)
            ]
            dw = 0.0
            # Compute STDP for each pair
            #consider the pre-post spike pair
            for t_pre in pre_times_step:
                for t_post in post_times_step:
                    #get the time difference
                    dt = (t_post - t_pre).magnitude
                    #If post-synaptic spike comes after pre-synaptic
                    if dt >= 5.0:
                        # Potentiation
                        dw += reward * A_plus * np.exp(-abs(dt-5.0) / tau_plus)
                    #If post fires before pre punish the connection.
                    elif dt <= -5.0:
                        # Depression
                        dw -= reward * A_minus * np.exp(-abs(dt+5.0) / tau_minus)

            # Update weight
            current_weights[pre_idx, post_idx] = np.clip(
                current_weights[pre_idx, post_idx] + dw,
                0.0, 1.0
            )
    return current_weights
```

```python
# -------------------------------
# Main Training Loop
# -------------------------------
if __name__ == "__main__":
    num_episodes = 500
    stats_interval = 10
    epsilon = 0.2
    epsilon_decay = 0.99
    episode_rewards = []

    # Tracking variables
    win_count = 0
    loss_count = 0
    draw_count = 0
    episode_stats = []
    episode_rewards = []
    
    env = Tictactoe_v0()
    # Initialize weights for both layers
#     input_hidden_weights = np.random.rand(27, 20) * 0.5
#     hidden_output_weights = np.random.rand(20, 9) * 0.5

    # Change weight initialization from 0-0.5 to 0.5-1.0
    input_hidden1_weights = np.random.uniform(1.0, 2.0, (27, 64))
    hidden1_hidden2_weights = np.random.uniform(1.0, 2.0, (64, 64))
    hidden2_output_weights = np.random.uniform(1.0, 2.0, (64, 9))


    # STDP parameters for the hidden-to-output layer (can be different)
    tau_plus_ho = 20.0
    tau_minus_ho = 20.0
    A_plus_ho = 0.05
    A_minus_ho = 0.05

#     # Initialize input->hidden weights in [0, 0.5]
#     current_weights = np.random.rand(27, 20) * 0.5

    for episode in range(num_episodes):
        state = env.reset(is_human_first=False)  # or False if you want AI to go first
        # AI goes second by playing as O. 
#         print(f"[DEBUG] AI is playing as {'X' if env.player_mark == 1 else 'O'}")
#         print(f"[DEBUG] AI goes {'first' if env.player_mark == 1 else 'second'}")
        total_reward = 0.0
        done = False

        # -----------------------
        # Build (and keep) one SNN for the entire episode
        # -----------------------
        sim.setup(timestep=10.0)
        run_offset = 0.0  # We'll accumulate time in ms

        # Create populations
        input_pop = sim.Population(27, sim.SpikeSourceArray(), label="Input")
        hidden1_pop = sim.Population(64, sim.IF_curr_exp(tau_m=15.0, v_thresh=-60.0, v_reset=-70.0), label="Hidden1")
        hidden2_pop = sim.Population(64, sim.IF_curr_exp(tau_m=15.0, v_thresh=-60.0, v_reset=-70.0), label="Hidden2")
        output_pop  = sim.Population(9, sim.IF_curr_exp(tau_m=10.0, v_thresh=-60.0, v_reset=-65.0), label="Output")

        # Create input->hidden (FromListConnector) with our current_weights
        input_hidden_conn = [
            (pre, post, input_hidden1_weights[pre, post], 5.0)
            for pre in range(27) for post in range(64)
        ]
        input_to_hidden = sim.Projection(
            input_pop, hidden1_pop,
            sim.FromListConnector(input_hidden_conn),
            receptor_type="excitatory",
        )

        hidden1_hidden2_conn = [
            (pre, post, hidden1_hidden2_weights[pre, post], 5.0)
            for pre in range(64) for post in range(64)
        ]
        hidden1_to_hidden2 = sim.Projection(
            hidden1_pop, hidden2_pop,
            sim.FromListConnector(hidden1_hidden2_conn),
            receptor_type="excitatory"
        )

        hidden2_output_conn = [
            (pre, post, hidden2_output_weights[pre, post], 5.0)
            for pre in range(64) for post in range(9)
        ]
        hidden_to_output = sim.Projection(
            hidden2_pop, output_pop,
            sim.FromListConnector(hidden2_output_conn),
            receptor_type="excitatory"
        )

        input_pop.record("spikes")
        hidden1_pop.record("spikes")
        hidden2_pop.record("spikes")
        output_pop.record(["spikes", "v"])


        # -----------------------
        # Step through the game
        # -----------------------
        while not done:
            valid_actions = env.get_valid_actions()
            step_duration = 30.0
            # If no valid actions => game must be done (draw).
            if not valid_actions:
                break

            # Encode state -> spike pattern
            encoded_state = population_encode(state)
            spike_times = [[] for _ in range(27)]
            for i in range(27):
                if encoded_state[i] > 0.5:
                    # Simple: spike once at the beginning of this step window
                    #spike_times[i] = [run_offset]
                    spike_times[i] = [run_offset + 1.0]  # within 0–30 ms
            # Update the SpikeSourceArray
            input_pop.set(spike_times=spike_times)

            # Run the simulation for 10 ms more
            step_duration = 15.0
            sim.run(step_duration)

            # Retrieve output spikes (so far)
            output_spike_data = output_pop.get_data("spikes").segments[-1].spiketrains
            action = decode_action_by_potential(output_pop, valid_actions)


            # Epsilon-greedy
            if (random.random() < epsilon) and valid_actions:
                # Random exploration
                action = random.choice(valid_actions)
            else:
                # If the decoded action isn't valid, override with a valid move
                if action not in valid_actions:
                    action = random.choice(valid_actions)

            # Make the move
            next_state, reward, done, _ = env.step(action)
            total_reward += reward
            
            # Perform STDP update
            input_spike_data = input_pop.get_data("spikes").segments[-1].spiketrains
            hidden1_spike_data = hidden1_pop.get_data("spikes").segments[-1].spiketrains
            hidden2_spike_data = hidden2_pop.get_data("spikes").segments[-1].spiketrains
            output_spike_data = output_pop.get_data("spikes").segments[-1].spiketrains

            input_hidden1_weights = stdp_per_synapse_update(
                input_hidden1_weights,
                reward,
                input_spike_data,
                hidden1_spike_data,
                run_offset=run_offset,
                step_size=step_duration
            )

            hidden1_hidden2_weights = stdp_per_synapse_update(
                hidden1_hidden2_weights,
                reward,
                hidden1_spike_data,
                hidden2_spike_data,
                run_offset=run_offset,
                step_size=step_duration
            )

            hidden2_output_weights = stdp_per_synapse_update(
                hidden2_output_weights,
                reward,
                hidden2_spike_data,
                output_spike_data,
                run_offset=run_offset,
                step_size=step_duration,
                tau_plus=tau_plus_ho,
                tau_minus=tau_minus_ho,
                A_plus=A_plus_ho,
                A_minus=A_minus_ho
            )



#             input_to_hidden.set(weights=input_hidden1_weights)
#             hidden1_to_hidden2.set(weights=hidden1_hidden2_weights)
#             hidden_to_output.set(weights=hidden2_output_weights)


            run_offset += step_duration
            state = next_state  # proceed
        
        # Update outcome counters based on the episode's total reward
        if total_reward > 0:
            win_count += 1
        elif total_reward < 0:
            loss_count += 1
        else:
            draw_count += 1
        
        
        # End SNN simulation for this episode
        sim.end()

        # Decay epsilon
        epsilon *= epsilon_decay
        episode_rewards.append(total_reward)
        print(f"Episode {episode+1}, Total Reward: {total_reward}")
    
        # Now, print the statistics only once per episode if the stats_interval condition is met
        if (episode + 1) % stats_interval == 0:
            total_games = win_count + loss_count + draw_count
            if total_games > 0:
                win_rate = win_count / total_games
                loss_rate = loss_count / total_games
                draw_rate = draw_count / total_games
            else:
                win_rate = loss_rate = draw_rate = 0
            
            episode_stats.append({
                'episodes': episode + 1,
                'win_rate': win_rate,
                'loss_rate': loss_rate,
                'draw_rate': draw_rate
            })
            
            print(f"\nAfter {episode+1} episodes:")
            print(f"Win rate: {win_rate:.2%}")
            print(f"Loss rate: {loss_rate:.2%}")
            print(f"Draw rate: {draw_rate:.2%}")

            # Reset counters after reporting if you want per-interval stats
            win_count = loss_count = draw_count = 0 
   

    # Final statistics
    print("\nFinal Statistics:")
    print(f"Total episodes: {num_episodes}")
    print(f"Overall win rate: {sum(s['win_rate'] for s in episode_stats)/len(episode_stats):.2%}")
    print(f"Overall loss rate: {sum(s['loss_rate'] for s in episode_stats)/len(episode_stats):.2%}")
    print(f"Overall draw rate: {sum(s['draw_rate'] for s in episode_stats)/len(episode_stats):.2%}")

    # Plotting
    plt.figure(figsize=(12, 6))
    episodes = [s['episodes'] for s in episode_stats]
    
    plt.plot(episodes, [s['win_rate'] for s in episode_stats], label='Win Rate')
    plt.plot(episodes, [s['loss_rate'] for s in episode_stats], label='Loss Rate')
    plt.plot(episodes, [s['draw_rate'] for s in episode_stats], label='Draw Rate')
    
    
    # Plot rewards
#     plt.plot(range(1, num_episodes+1), episode_rewards, marker='o')
    plt.xlabel("Episode")
    plt.ylabel("Overall Performance")
    plt.title("Performance rate Over Episodes")
    plt.show()
```
